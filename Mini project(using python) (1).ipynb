```python
df_num=df.select_dtypes(include=["number"])
df_cat=df.select_dtypes(exclude=["number"])
```

```python
for column in df_num.columns:
    print(column)
```

```python
df_num.isnull().sum()
```

```python
for column in df_num.columns:
   df_num[column]=df_num[column].fillna(df_num[column].median())
```

```python
df_num.isnull().sum()
```

```python
df_cat["buy_box_availability"].mode().values[0]
```

```python
df_cat["is_sponsored"].mode().values[0]
```

```python
for column in df_cat:
   mode=(df_cat[column].mode().values[0])
   df_cat[column]=df_cat[column].fillna(mode)
```

```python
df_cat.isnull().sum()
```

```python
df=pd.concat([df_num,df_cat],axis=1)
df.head()
```

```python
df.isnull().sum()
```

```python
df["total_reviews"]=df["total_reviews"].astype(int)
df["purchased_last_month"]=df["purchased_last_month"].astype(int)
#df["delivery_date"]=df["delivery_date"].astype(date)
```

```python
df.info()
```

```python
df["delivery_date"]=pd.to_datetime(df["delivery_date"],errors="coerce")
df["data_collected_at"]=pd.to_datetime(df["data_collected_at"],errors="coerce")
```

```python
df.info()
```

```python
df.drop(["sustainability_tags"],axis=1,inplace=True)
```

```python
df.info()
```

# Feature Engineering


```python
df["discounted_amount"]=df["original_price"]-df["discounted_price"]
```

```python
df["estimated_revenue"]=df["discounted_price"]*df["purchased_last_month"]
```

```python
df["percentage_band"]=pd.cut(x=df["discount_percentage"],bins=[0,10,30,70,99],labels=["less discount","medium discount","High","very high"])
```
